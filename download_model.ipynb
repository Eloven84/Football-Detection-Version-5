```python
import requests
import base64
import cv2

def infer_roboflow(frame, api_key, model_id="football-field-detection-f07vi/14"):
    _, buffer = cv2.imencode(".jpg", frame)
    img_b64 = base64.b64encode(buffer).decode("utf-8")
    
    url = f"https://detect.roboflow.com/{model_id}"
    response = requests.post(
        url,
        params={"api_key": api_key},
        data=img_b64,
        headers={"Content-Type": "application/x-www-form-urlencoded"}
    )
    return response.json()

frame = cv2.imread("debug_keypoints_frame683.jpg")
result = infer_roboflow(frame, api_key="z5JXv6j0AT4yg2hwaKGq")
print(result)
```

```python
import cv2
import numpy as np
from pitch_config_roboflow import PITCH_KEYPOINTS

# Simulasi dengan keypoint yang sering muncul di log
# Ambil dari debug_keypoints_frame100.jpg yang punya 17 KP
# Ganti koordinat ini dengan hasil aktual dari all_keypoints_dict[100]

# Load keypoints frame 100 dari pickle atau jalankan detector sekali
# Sementara test dengan frame yang kita tahu ada banyak KP

from keypoint_detector import KeypointDetector
from utils import read_video

video_frames = read_video('input_video/08fd33_4.mp4')
detector = KeypointDetector(
    model_path='models/best_roboflow_keypoints_detection.pt',
    confidence_threshold=0.3
)

# Test hanya frame 100 (yang punya 17 KP)
import cv2
result = detector.detect_keypoints(video_frames[100])
print("Frame 100 keypoints:", result)

# Hitung H manual
common_ids = [k for k in result if k in PITCH_KEYPOINTS]
print(f"Common IDs: {common_ids}")

src_pts = np.array([result[k] for k in common_ids], dtype=np.float32)
dst_pts = np.array([PITCH_KEYPOINTS[k] for k in common_ids], dtype=np.float32)
H, mask = cv2.findHomography(src_pts, dst_pts, cv2.RANSAC, 8.0)
# hitung reproj error...

if H is not None:
    src_h = np.hstack([src_pts, np.ones((len(src_pts), 1))])
    proj = (H @ src_h.T).T
    proj = proj[:, :2] / proj[:, 2:3]
    err = np.mean(np.linalg.norm(proj - dst_pts, axis=1))
    print(f"Reprojection error: {err:.2f}")
else:
    print("RANSAC failed")
```

```python
import cv2
import numpy as np
from ultralytics import YOLO

# ── config ──────────────────────────────────────────────────────────────
VIDEO_PATH  = "input_video/08fd33_4.mp4"
MODEL_PATH  = "models/best_roboflow_keypoints_detection.pt"  # sesuaikan path model lokal kamu
TEST_FRAMES = [0, 100, 300, 500, 650, 683, 700]
CONF_THRESH = 0.3
# ────────────────────────────────────────────────────────────────────────

model = YOLO(MODEL_PATH)
cap   = cv2.VideoCapture(VIDEO_PATH)

for fn in TEST_FRAMES:
    cap.set(cv2.CAP_PROP_POS_FRAMES, fn)
    ret, frame = cap.read()
    if not ret:
        print(f"Frame {fn:4d}: could not read")
        continue

    results = model(frame, verbose=False)[0]

    # keypoints ada di results.keypoints kalau ini pose/keypoint model
    if results.keypoints is None or len(results.keypoints) == 0:
        print(f"Frame {fn:4d}: no keypoints detected")
        continue

    kps_xy   = results.keypoints.xy[0].cpu().numpy()    # shape (N, 2)
    kps_conf = results.keypoints.conf[0].cpu().numpy()  # shape (N,)

    total     = len(kps_conf)
    high_conf = int((kps_conf >= CONF_THRESH).sum())

    print(f"\nFrame {fn:4d} | total_kp={total} | kp_conf>={CONF_THRESH}: {high_conf}")
    print(f"  {'ID':>3}  {'x':>6}  {'y':>6}  {'conf':>6}  {'ok?'}")
    print(f"  {'-'*38}")

    for kid in range(total):
        x, y = kps_xy[kid]
        c    = kps_conf[kid]
        flag = "✓" if c >= CONF_THRESH else "✗ LOW"
        print(f"  {kid:>3}  {x:>6.1f}  {y:>6.1f}  {c:>6.3f}  {flag}")

    print(f"  → conf stats: min={kps_conf.min():.3f}  "
          f"mean={kps_conf.mean():.3f}  max={kps_conf.max():.3f}")

cap.release()
print("\nDone.")
```

```python
# cek_keypoint_mapping.py
# Jalankan ini, screenshot outputnya dan share

from pitch_config_roboflow import PITCH_KEYPOINTS
import cv2
import numpy as np

CANVAS_W, CANVAS_H = 1050, 680

# Gambar lapangan
img = np.zeros((CANVAS_H, CANVAS_W, 3), dtype=np.uint8)
img[:] = (34, 139, 34)

MARGIN_X, MARGIN_Y = 50, 50
FIELD_W, FIELD_H = 950, 580

def m_to_px(x_m, y_m):
    px = MARGIN_X + int(x_m / 105.0 * FIELD_W)
    py = MARGIN_Y + int(y_m / 68.0 * FIELD_H)
    return (px, py)

# Garis lapangan
cv2.rectangle(img, (MARGIN_X, MARGIN_Y),
              (MARGIN_X+FIELD_W, MARGIN_Y+FIELD_H), (255,255,255), 2)
cv2.line(img, m_to_px(52.5, 0), m_to_px(52.5, 68), (255,255,255), 2)
cx, cy = m_to_px(52.5, 34)
cv2.circle(img, (cx,cy), int(9.15/105*FIELD_W), (255,255,255), 2)

# Plot SEMUA keypoints dengan ID yang besar + warna berbeda per zona
FOCUS_IDS = [13, 14, 15, 16, 30, 31, 17, 18, 9]  # yang sering conf tinggi

for kid, (px, py) in PITCH_KEYPOINTS.items():
    if kid in FOCUS_IDS:
        color = (0, 255, 255)   # kuning = focus
        r = 12
    else:
        color = (150, 150, 150) # abu = lainnya
        r = 6
    cv2.circle(img, (px, py), r, color, -1)
    cv2.putText(img, str(kid), (px+5, py-5),
                cv2.FONT_HERSHEY_SIMPLEX, 0.5,
                (255,255,0) if kid in FOCUS_IDS else (200,200,200), 1)

cv2.imwrite("debug_kp_mapping.jpg", img)
print("Saved: debug_kp_mapping.jpg")
print("\nKoordinat PITCH_KEYPOINTS untuk ID yang sering terdeteksi:")
for kid in FOCUS_IDS:
    if kid in PITCH_KEYPOINTS:
        px, py = PITCH_KEYPOINTS[kid]
        print(f"  ID {kid:2d}: canvas=({px:4d}, {py:4d})")
```
