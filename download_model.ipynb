```python
import requests
import base64
import cv2

def infer_roboflow(frame, api_key, model_id="football-field-detection-f07vi/14"):
    _, buffer = cv2.imencode(".jpg", frame)
    img_b64 = base64.b64encode(buffer).decode("utf-8")
    
    url = f"https://detect.roboflow.com/{model_id}"
    response = requests.post(
        url,
        params={"api_key": api_key},
        data=img_b64,
        headers={"Content-Type": "application/x-www-form-urlencoded"}
    )
    return response.json()

frame = cv2.imread("debug_keypoints_frame683.jpg")
result = infer_roboflow(frame, api_key="z5JXv6j0AT4yg2hwaKGq")
print(result)
```

```python
import cv2
import base64
import requests

# Pakai frame 650 yang tadi
cap = cv2.VideoCapture("input_video/08fd33_4.mp4")
cap.set(cv2.CAP_PROP_POS_FRAMES, 650)
ret, frame = cap.read()
cap.release()

# Infer
_, buffer = cv2.imencode(".jpg", frame)
img_b64 = base64.b64encode(buffer).decode("utf-8")
response = requests.post(
    "https://detect.roboflow.com/football-field-detection-f07vi/14",
    params={"api_key": "z5JXv6j0AT4yg2hwaKGq"},
    data=img_b64,
    headers={"Content-Type": "application/x-www-form-urlencoded"}
)
result = response.json()

# Visualisasi semua keypoint dengan label ID
debug = frame.copy()
for kp in result["predictions"][0]["keypoints"]:
    x, y, conf, kid = int(kp["x"]), int(kp["y"]), kp["confidence"], kp["class_id"]
    if conf < 0.3 or x <= 0:
        continue
    cv2.circle(debug, (x, y), 8, (0, 255, 0), -1)
    cv2.putText(debug, str(kid), (x + 6, y - 6),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 255), 2)

cv2.imwrite("debug_roboflow_kp_frame650.jpg", debug)
print("Saved debug_roboflow_kp_frame650.jpg")
```

```python
# Cek nama class tiap keypoint dari API response
for kp in result["predictions"][0]["keypoints"]:
    if kp["confidence"] > 0.3 and kp["x"] > 0:
        print(f"kp {kp['class_id']:2d} | class: {kp['class']:>3} | "
              f"x={kp['x']:.0f} y={kp['y']:.0f} conf={kp['confidence']:.3f}")
```

```python
import cv2, base64, requests

api_key = "z5JXv6j0AT4yg2hwaKGq"
test_frames = [0, 100, 300, 500, 650, 683, 700]

cap = cv2.VideoCapture("input_video/08fd33_4.mp4")

for fn in test_frames:
    cap.set(cv2.CAP_PROP_POS_FRAMES, fn)
    ret, frame = cap.read()
    if not ret:
        continue

    _, buffer = cv2.imencode(".jpg", frame)
    img_b64 = base64.b64encode(buffer).decode("utf-8")
    response = requests.post(
        "https://detect.roboflow.com/football-field-detection-f07vi/14",
        params={"api_key": api_key},
        data=img_b64,
        headers={"Content-Type": "application/x-www-form-urlencoded"}
    )
    result = response.json()
    preds = result.get("predictions", [])
    if not preds:
        print(f"Frame {fn:4d}: no prediction")
        continue

    kps = {kp["class_id"]: kp for kp in preds[0]["keypoints"]}
    print(f"\nFrame {fn:4d}:")
    for kid in [27, 28, 29, 30, 31]:
        kp = kps.get(kid)
        if kp:
            print(f"  kp {kid}: x={kp['x']:.0f} y={kp['y']:.0f} conf={kp['confidence']:.3f}")
        else:
            print(f"  kp {kid}: not in response")

cap.release()
```
